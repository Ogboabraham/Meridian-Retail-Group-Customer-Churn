 The Meridian Retail Group customer churn project uses five operational datasets: Customers, Products, Orders, Order Items, and Support Tickets. The datasets were loaded from the project's Raw Data folder stored in Google Drive and imported into the Google Colab environment using Python and Pandas.

 The data-type inspection was performed using Pandas to identify numerical, categorical, date/time and other fields. This initial assessment provides the foundation for subsequent data-quality checks and preparation.

2.2 Entity-Relationship Structure

The five datasets represent interconnected business entities. Customers are connected to Orders through the customer identifier. Orders are connected to Order Items through the order identifier, while Order Items connect to Products through the product identifier. Customers are also connected to Support Tickets through the customer identifier.

This structure allows the analysis to combine customer information, purchasing behaviour, product information and customer-service interactions when investigating customer churn.

```python
2.3 Business Origin of the Datasets

Dataset| Plausible Operational Source| Business Purpose
Customers| CRM / Customer Master System| Stores customer account and profile information
Products| Product / Inventory Management System| Stores information about products sold by Meridian
Orders| E-commerce, POS or Order Management System| Records customer orders and transactions
Order Items| Order Management / Transaction System| Records the individual products and quantities contained in each order
Support Tickets| Customer Service / Help-Desk / CRM System| Records customer enquiries, complaints and support interactions

These datasets were acquired as raw operational data and will subsequently be examined for missing values, duplicate records, inconsistent formats, incorrect data types and other quality issues during the data preparation stage.
```
