```python
# Calculating IQR outlier counts
def outlier_summary(df):
    numerical_columns = df.select_dtypes(include="number").columns

    results = []

    for column in numerical_columns:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[
            (df[column] < lower_bound) |
            (df[column] > upper_bound)
        ]

        results.append({
            "Column": column,
            "Q1": Q1,
            "Q3": Q3,
            "IQR": IQR,
            "Lower Bound": lower_bound,
            "Upper Bound": upper_bound,
            "Outlier Count": len(outliers)
        })

    return pd.DataFrame(results)
```

```python
print("========== CUSTOMERS OUTLIERS ==========")
display(outlier_summary(customers))

print("========== PRODUCTS OUTLIERS ==========")
display(outlier_summary(products))

print("========== ORDERS OUTLIERS ==========")
display(outlier_summary(orders))

print("========== ORDER ITEMS OUTLIERS ==========")
display(outlier_summary(order_items))
print("========== SUPPORT TICKETS OUTLIERS ==========")
display(outlier_summary(support_tickets))
```

 ***SOME OF THE FINDINGS INCLUDES***

1. Customers

Date of birth

1899-01-01 appears to be a placeholder/invalid DOB because those records also have missing Age.
2031-05-04 is a future date relative to the dataset period and is therefore invalid.
The normal DOBs appear reasonable.

Decision: Convert date_of_birth to datetime and replace clearly invalid placeholder/future dates with missing (NaT). We should not invent replacement dates.

Revenue outliers

The highest customer revenue is ₦35,395.37 with 14 orders.
The outlier customers have substantial order activity; they don't look like obvious data-entry errors.
Therefore, these are likely genuine high-value customers, which are especially important in a churn investigation.

Decision: Keep revenue outliers. Do not cap or remove them.

2. Products

The four unit_cost outliers are high, but importantly, they are still plausible positive monetary values.

However, there is an important issue in the data:

For some products, unit_cost is greater than unit_price.
For example, PROD1037 has unit cost 1019.355 but unit price 679.57.

That is potentially a business/data-quality issue, rather than simply a statistical outlier.

Decision for now: Don't remove these products merely because they are IQR outliers. We should preserve the observations and flag the cost-price inconsistencies for investigation rather than destroying potentially useful records.

3. Orders

The discount results are actually very good:

0, 5, 10, 15, 20, 25

All are sensible percentage values, and the maximum is only 25%.

Decision: Keep all discount values. Although IQR identified some as statistical outliers, they are valid business values and should not be removed.

The high total_amount values should also be retained. A large order is not automatically an error.

Decision: Keep transaction-value outliers unless we find evidence that they are impossible or incorrectly entered.

4. Order Items

This is the most important outlier problem we found.

quantity has:

Minimum = -4
Maximum = 4
Normal quantities appear to be 1–4.

A negative quantity is not a valid ordinary purchase quantity.

Decision: Negative quantities should be treated as invalid data, not as statistical outliers.

We should identify exactly how many negative quantities exist and decide whether they represent returns/cancellations or data-entry errors before removing them.

The positive quantity of 4 is perfectly reasonable, so we should not remove quantity = 4.

5. Duplicate records

We now have very strong evidence that the duplicate records are genuine exact duplicates.

For example:

ORD102482 appears twice with exactly the same customer, date, channel, status, payment method, shipping cost, discount and amount.

Similarly, the order-item duplicates have exactly the same:

order_item_id + order_id + product_id + quantity + unit_price_at_purchase.

Therefore:

Orders: Remove exact duplicate rows — 25 duplicate rows.

Order Items: Remove exact duplicate rows — 40 duplicate rows.

We should not remove rows simply because order_id repeats in order_items, because one order can legitimately contain multiple products/line items.

** IN CONCLUSION **

The data cleaning and transformation process was conducted systematically across the Customers, Products, Orders, and Order Items datasets. Decisions were based on the nature of each data-quality issue and its potential effect on the subsequent customer churn analysis. Missing values were retained where no reliable replacement could be determined, while selected variables were imputed where a defensible business-based estimate was available. Invalid dates, inconsistent categorical values, duplicate records, and inappropriate data types were addressed to improve data consistency and analytical reliability.

Statistical outliers were not automatically removed. Where extreme values represented plausible business activity, such as high customer revenue or high-value orders, they were retained because they may contain important information about customer value and churn behaviour. Conversely, logically invalid observations, such as negative quantities that could not be associated with returned orders, were treated as data-quality errors.

Exact duplicate records were removed from the Orders and Order Items datasets because they represented repeated copies of the same observations. However, repeated order IDs within the Order Items dataset were retained because one order may legitimately contain multiple line items.

Feature scaling was deliberately excluded from the cleaned datasets. Scaling will be applied during the modeling stage only where required by the selected machine-learning algorithms. This approach preserves the original business meaning of the cleaned data while ensuring that model-specific preprocessing is performed appropriately.

```python
# ============================================================
# PHASE 6 — DATA CLEANING & TRANSFORMATION
# Meridian Retail Group
# ============================================================

import pandas as pd
import numpy as np
import os

# ------------------------------------------------------------
# 1. MOUNT GOOGLE DRIVE
# ------------------------------------------------------------

from google.colab import drive
drive.mount('/content/drive')


# ------------------------------------------------------------
# 2. DEFINE FOLDERS
# ------------------------------------------------------------

processed_path = "/content/drive/MyDrive/Meridian Retail Group - Customer Churn Capstone /Data/Processed Data"
cleaned_path = "/content/drive/MyDrive/Meridian Retail Group - Customer Churn Capstone /Data/Cleaned Data"

# Create Cleaned Data folder if it does not exist
os.makedirs(cleaned_path, exist_ok=True)

print("Processed Data folder:")
print(processed_path)

print("\nCleaned Data folder:")
print(cleaned_path)




# ------------------------------------------------------------
# 3. LOAD PHASE 5 PROCESSED DATA
# ------------------------------------------------------------

customers = pd.read_csv(
    os.path.join(processed_path, "customers_processed.csv")
)

products = pd.read_csv(
    os.path.join(processed_path, "products_processed.csv")
)

orders = pd.read_csv(
    os.path.join(processed_path, "orders_processed.csv")
)

order_items = pd.read_csv(
    os.path.join(processed_path, "order_items_processed.csv")
)

support_tickets = pd.read_csv(
    os.path.join(processed_path, "support_tickets_processed.csv")
)

print("\nDatasets loaded successfully.")

print("\nDataset shapes:")
print("Customers:", customers.shape)
print("Products:", products.shape)
print("Orders:", orders.shape)
print("Order Items:", order_items.shape)
print("Support Tickets:", support_tickets.shape)



# ============================================================
# 4. CUSTOMERS CLEANING
# ============================================================

print("\n" + "=" * 70)
print("CLEANING CUSTOMERS")
print("=" * 70)

# Make a copy so the original loaded dataframe is preserved
customers_clean = customers.copy()


# ------------------------------------------------------------
# 4.1 Convert date columns safely
# ------------------------------------------------------------

customers_clean["date_of_birth"] = pd.to_datetime(
    customers_clean["date_of_birth"],
    errors="coerce"
)

customers_clean["signup_date"] = pd.to_datetime(
    customers_clean["signup_date"],
    errors="coerce"
)


# ------------------------------------------------------------
# 4.2 Remove clearly invalid DOB values
# ------------------------------------------------------------

# 1899-01-01 appears to be a placeholder date
customers_clean.loc[
    customers_clean["date_of_birth"] == pd.Timestamp("1899-01-01"),
    "date_of_birth"
] = pd.NaT

# Future birth dates are invalid
customers_clean.loc[
    customers_clean["date_of_birth"] > pd.Timestamp("2026-09-09"),
    "date_of_birth"
] = pd.NaT


# ------------------------------------------------------------
# 4.3 Standardize gender
# ------------------------------------------------------------

customers_clean["gender"] = (
    customers_clean["gender"]
    .astype("string")
    .str.strip()
    .str.title()
)

customers_clean["gender"] = customers_clean["gender"].fillna("Unknown")


# ------------------------------------------------------------
# 4.4 Standardize membership tier
# ------------------------------------------------------------

customers_clean["membership_tier"] = (
    customers_clean["membership_tier"]
    .astype("string")
    .str.strip()
    .str.title()
)

customers_clean["membership_tier"] = (
    customers_clean["membership_tier"]
    .fillna("Unknown")
)


# ------------------------------------------------------------
# 4.5 Standardize marketing opt-in
# ------------------------------------------------------------

marketing_map = {
    "Y": "Yes",
    "y": "Yes",
    "YES": "Yes",
    "Yes": "Yes",
    "1": "Yes",
    "1.0": "Yes",

    "N": "No",
    "n": "No",
    "NO": "No",
    "No": "No",
    "0": "No",
    "0.0": "No"
}

customers_clean["marketing_opt_in"] = (
    customers_clean["marketing_opt_in"]
    .astype("string")
    .str.strip()
    .replace(marketing_map)
    .fillna("Unknown")
)


# ------------------------------------------------------------
# 4.6 Standardize state names
# ------------------------------------------------------------

state_map = {
    "NY": "New York",
    "AZ": "Arizona",
    "OH": "Ohio",
    "GA": "Georgia",
    "TX": "Texas",
    "MA": "Massachusetts",
    "PA": "Pennsylvania"
}

customers_clean["state"] = (
    customers_clean["state"]
    .astype("string")
    .str.strip()
    .replace(state_map)
)

customers_clean["state"] = customers_clean["state"].fillna("Unknown")


# ------------------------------------------------------------
# 4.7 Standardize referral source
# ------------------------------------------------------------

customers_clean["referral_source"] = (
    customers_clean["referral_source"]
    .astype("string")
    .str.strip()
    .fillna("Unknown")
)


# ------------------------------------------------------------
# 4.8 Standardize city
# ------------------------------------------------------------

customers_clean["city"] = (
    customers_clean["city"]
    .astype("string")
    .str.strip()
    .fillna("Unknown")
)


# ------------------------------------------------------------
# 4.9 Standardize preferred channel
# ------------------------------------------------------------

customers_clean["preferred_channel"] = (
    customers_clean["preferred_channel"]
    .astype("string")
    .str.strip()
    .str.title()
    .fillna("Unknown")
)


# ------------------------------------------------------------
# 4.10 Handle missing satisfaction
# ------------------------------------------------------------

# Missing satisfaction is retained and flagged rather than
# blindly replacing it with an average.

customers_clean["Satisfaction_Missing"] = (
    customers_clean["Average_Satisfaction"].isna().astype(int)
)


# ------------------------------------------------------------
# 4.11 Remove exact duplicate customer records
# ------------------------------------------------------------

customers_clean = customers_clean.drop_duplicates()


# ============================================================
# 5. PRODUCTS CLEANING
# ============================================================

print("\n" + "=" * 70)
print("CLEANING PRODUCTS")
print("=" * 70)

products_clean = products.copy()


# ------------------------------------------------------------
# 5.1 Convert launch date
# ------------------------------------------------------------

products_clean["launch_date"] = pd.to_datetime(
    products_clean["launch_date"],
    errors="coerce"
)


# ------------------------------------------------------------
# 5.2 Standardize categorical values
# ------------------------------------------------------------

for col in ["category", "sub_category", "brand", "is_active", "Price_Level"]:

    products_clean[col] = (
        products_clean[col]
        .astype("string")
        .str.strip()
    )


# ------------------------------------------------------------
# 5.3 Handle missing sub-category
# ------------------------------------------------------------

products_clean["sub_category"] = (
    products_clean["sub_category"]
    .fillna("Unknown")
)


# ------------------------------------------------------------
# 5.4 Handle missing unit price
# ------------------------------------------------------------

# Use category-level median because products within the same
# category are more comparable than using one global median.

category_median_price = (
    products_clean.groupby("category")["unit_price"]
    .transform("median")
)

products_clean["unit_price"] = (
    products_clean["unit_price"]
    .fillna(category_median_price)
)


# ------------------------------------------------------------
# 5.5 Create a cost-price inconsistency flag
# ------------------------------------------------------------

products_clean["Cost_Greater_Than_Price"] = (
    products_clean["unit_cost"] > products_clean["unit_price"]
).astype(int)


# ------------------------------------------------------------
# 5.6 Remove exact duplicate product records
# ------------------------------------------------------------

products_clean = products_clean.drop_duplicates()


# ============================================================
# 6. ORDERS CLEANING
# ============================================================

print("\n" + "=" * 70)
print("CLEANING ORDERS")
print("=" * 70)

orders_clean = orders.copy()


# ------------------------------------------------------------
# 6.1 Convert order date
# ------------------------------------------------------------

orders_clean["order_date"] = pd.to_datetime(
    orders_clean["order_date"],
    errors="coerce"
)


# ------------------------------------------------------------
# 6.2 Standardize channel
# ------------------------------------------------------------

orders_clean["channel"] = (
    orders_clean["channel"]
    .astype("string")
    .str.strip()
    .str.title()
)


# ------------------------------------------------------------
# 6.3 Standardize order status
# ------------------------------------------------------------

orders_clean["order_status"] = (
    orders_clean["order_status"]
    .astype("string")
    .str.strip()
    .str.title()
)


# ------------------------------------------------------------
# 6.4 Standardize payment method
# ------------------------------------------------------------

orders_clean["payment_method"] = (
    orders_clean["payment_method"]
    .astype("string")
    .str.strip()
    .str.title()
)


# ------------------------------------------------------------
# 6.5 Ensure numeric columns have correct data types
# ------------------------------------------------------------

orders_clean["shipping_cost"] = pd.to_numeric(
    orders_clean["shipping_cost"],
    errors="coerce"
)

orders_clean["discount_pct"] = pd.to_numeric(
    orders_clean["discount_pct"],
    errors="coerce"
)

orders_clean["total_amount"] = pd.to_numeric(
    orders_clean["total_amount"],
    errors="coerce"
)


# ------------------------------------------------------------
# 6.6 Remove exact duplicate order records
# ------------------------------------------------------------

before_orders = len(orders_clean)

orders_clean = orders_clean.drop_duplicates()

after_orders = len(orders_clean)

print(f"Exact duplicate order rows removed: {before_orders - after_orders}")


# ------------------------------------------------------------
# 6.7 Keep missing order dates
# ------------------------------------------------------------

# We do NOT invent dates.
# Missing dates will remain NaT.


# ============================================================
# 7. ORDER ITEMS CLEANING
# ============================================================

print("\n" + "=" * 70)
print("CLEANING ORDER ITEMS")
print("=" * 70)

order_items_clean = order_items.copy()


# ------------------------------------------------------------
# 7.1 Ensure numeric columns have correct types
# ------------------------------------------------------------

order_items_clean["quantity"] = pd.to_numeric(
    order_items_clean["quantity"],
    errors="coerce"
)

order_items_clean["unit_price_at_purchase"] = pd.to_numeric(
    order_items_clean["unit_price_at_purchase"],
    errors="coerce"
)


# ------------------------------------------------------------
# 7.2 Identify negative quantities
# ------------------------------------------------------------

negative_quantity_count = (
    order_items_clean["quantity"] < 0
).sum()

print(
    f"Negative quantity records found: "
    f"{negative_quantity_count}"
)


# ------------------------------------------------------------
# 7.3 Check whether negative quantities are associated
#     with returned orders
# ------------------------------------------------------------

returned_orders = set(
    orders_clean.loc[
        orders_clean["order_status"] == "Returned",
        "order_id"
    ]
)

negative_quantity_rows = order_items_clean[
    order_items_clean["quantity"] < 0
].copy()

negative_quantity_rows["Associated_With_Return"] = (
    negative_quantity_rows["order_id"]
    .isin(returned_orders)
)

print("\nNegative quantities associated with returned orders:")
print(
    negative_quantity_rows[
        "Associated_With_Return"
    ].value_counts(dropna=False)
)


# ------------------------------------------------------------
# 7.4 Handle negative quantities
# ------------------------------------------------------------

# Negative quantities associated with returned orders may
# represent legitimate returns, so they are retained.
#
# Negative quantities NOT associated with returned orders
# are treated as invalid records and removed.

invalid_negative_rows = (
    (order_items_clean["quantity"] < 0)
    &
    (~order_items_clean["order_id"].isin(returned_orders))
)

removed_negative = invalid_negative_rows.sum()

order_items_clean = order_items_clean[
    ~invalid_negative_rows
].copy()

print(
    f"Invalid negative quantity rows removed: "
    f"{removed_negative}"
)


# ------------------------------------------------------------
# 7.5 Handle missing purchase prices
# ------------------------------------------------------------

# We use the product's cleaned unit price when the
# purchase price is missing.

product_price_lookup = (
    products_clean
    .set_index("product_id")["unit_price"]
)

missing_purchase_price = (
    order_items_clean["unit_price_at_purchase"].isna()
)

order_items_clean.loc[
    missing_purchase_price,
    "unit_price_at_purchase"
] = (
    order_items_clean.loc[
        missing_purchase_price,
        "product_id"
    ].map(product_price_lookup)
)


# ------------------------------------------------------------
# 7.6 Remove exact duplicate order-item records
# ------------------------------------------------------------

before_items = len(order_items_clean)

order_items_clean = order_items_clean.drop_duplicates()

after_items = len(order_items_clean)

print(
    f"Exact duplicate order-item rows removed: "
    f"{before_items - after_items}"
)


# ============================================================
# 8. SUPPORT TICKETS CLEANING
# ============================================================

print("\n" + "=" * 70)
print("CLEANING SUPPORT TICKETS")
print("=" * 70)

support_tickets_clean = support_tickets.copy()


# ------------------------------------------------------------
# 8.1 Display available columns
# ------------------------------------------------------------

print("\nSupport ticket columns:")
print(list(support_tickets_clean.columns))


# ------------------------------------------------------------
# 8.2 Remove exact duplicate ticket records
# ------------------------------------------------------------

before_tickets = len(support_tickets_clean)

support_tickets_clean = (
    support_tickets_clean
    .drop_duplicates()
    .copy()
)

after_tickets = len(support_tickets_clean)

print(
    f"\nExact duplicate support-ticket rows removed: "
    f"{before_tickets - after_tickets}"
)


# ------------------------------------------------------------
# 8.3 Standardize text columns
# ------------------------------------------------------------

# Clean all object/string columns by removing unnecessary
# leading/trailing spaces.

for col in support_tickets_clean.select_dtypes(
    include=["object", "string"]
).columns:

    support_tickets_clean[col] = (
        support_tickets_clean[col]
        .astype("string")
        .str.strip()
    )


# ------------------------------------------------------------
# 8.4 Standardize ticket status
# ------------------------------------------------------------

# If a status column exists, standardize its formatting.

status_candidates = [
    "status",
    "ticket_status",
    "Status",
    "Ticket_Status"
]

status_column = next(
    (
        col
        for col in status_candidates
        if col in support_tickets_clean.columns
    ),
    None
)

if status_column is not None:

    support_tickets_clean[status_column] = (
        support_tickets_clean[status_column]
        .astype("string")
        .str.strip()
        .str.title()
    )

    print(
        f"\nStatus column standardized: "
        f"{status_column}"
    )

else:

    print("\nNo ticket status column detected.")


# ------------------------------------------------------------
# 8.5 Convert satisfaction score to numeric
# ------------------------------------------------------------

satisfaction_candidates = [
    "satisfaction_score",
    "satisfaction",
    "Satisfaction_Score",
    "Satisfaction"
]

satisfaction_column = next(
    (
        col
        for col in satisfaction_candidates
        if col in support_tickets_clean.columns
    ),
    None
)

if satisfaction_column is not None:

    support_tickets_clean[satisfaction_column] = pd.to_numeric(
        support_tickets_clean[satisfaction_column],
        errors="coerce"
    )

    print(
        f"Satisfaction column converted to numeric: "
        f"{satisfaction_column}"
    )

else:

    print("No satisfaction column detected.")


# ------------------------------------------------------------
# 8.6 Convert resolution time to numeric
# ------------------------------------------------------------

resolution_candidates = [
    "resolution_time",
    "resolution_time_hours",
    "resolution_hours",
    "Resolution_Time",
    "Resolution_Time_Hours"
]

resolution_column = next(
    (
        col
        for col in resolution_candidates
        if col in support_tickets_clean.columns
    ),
    None
)

if resolution_column is not None:

    support_tickets_clean[resolution_column] = pd.to_numeric(
        support_tickets_clean[resolution_column],
        errors="coerce"
    )

    print(
        f"Resolution-time column converted to numeric: "
        f"{resolution_column}"
    )

else:

    print("No resolution-time column detected.")


# ------------------------------------------------------------
# 8.7 Convert date columns safely
# ------------------------------------------------------------

date_candidates = [
    "created_date",
    "ticket_date",
    "opened_date",
    "closed_date",
    "resolved_date",
    "created_at",
    "closed_at",
    "resolved_at",
    "Created_Date",
    "Ticket_Date",
    "Opened_Date",
    "Closed_Date",
    "Resolved_Date"
]

for col in date_candidates:

    if col in support_tickets_clean.columns:

        support_tickets_clean[col] = pd.to_datetime(
            support_tickets_clean[col],
            errors="coerce"
        )

        print(
            f"Date column converted: {col}"
        )


# ------------------------------------------------------------
# 8.8 Check negative resolution times
# ------------------------------------------------------------

if resolution_column is not None:

    negative_resolution = (
        support_tickets_clean[resolution_column] < 0
    ).sum()

    print(
        f"\nNegative resolution-time records found: "
        f"{negative_resolution}"
    )

    # Negative resolution time cannot be logically valid.
    # Therefore, convert these values to missing.

    support_tickets_clean.loc[
        support_tickets_clean[resolution_column] < 0,
        resolution_column
    ] = np.nan

    print(
        "Negative resolution times converted to missing values."
    )


# ------------------------------------------------------------
# 8.9 Validate satisfaction scores
# ------------------------------------------------------------

if satisfaction_column is not None:

    print("\nSatisfaction score summary:")
    print(
        support_tickets_clean[
            satisfaction_column
        ].describe()
    )

    # Satisfaction scores outside the normal 1–5 range
    # are treated as invalid.

    invalid_satisfaction = (
        (support_tickets_clean[satisfaction_column] < 1)
        |
        (support_tickets_clean[satisfaction_column] > 5)
    )

    invalid_satisfaction_count = (
        invalid_satisfaction.sum()
    )

    print(
        f"Invalid satisfaction scores found: "
        f"{invalid_satisfaction_count}"
    )

    support_tickets_clean.loc[
        invalid_satisfaction,
        satisfaction_column
    ] = np.nan

    print(
        "Invalid satisfaction scores converted to missing."
    )


# ------------------------------------------------------------
# 8.10 Validate customer IDs
# ------------------------------------------------------------

if "customer_id" in support_tickets_clean.columns:

    missing_customer_ids = (
        support_tickets_clean["customer_id"].isna()
    ).sum()

    print(
        f"\nMissing customer IDs: "
        f"{missing_customer_ids}"
    )


# ------------------------------------------------------------
# 8.11 Final support-ticket information
# ------------------------------------------------------------

print("\nSupport Tickets after cleaning:")
print(
    "Shape:",
    support_tickets_clean.shape
)

print("\nMissing values:")
print(
    support_tickets_clean.isna().sum()
)

print("\nData types:")
print(
    support_tickets_clean.dtypes
)



# ============================================================
# 9. FINAL VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("FINAL VALIDATION")
print("=" * 70)


# ------------------------------------------------------------
# Customers
# ------------------------------------------------------------

print("\nCUSTOMERS")
print("Shape:", customers_clean.shape)
print("Exact duplicates:", customers_clean.duplicated().sum())
print("\nMissing values:")
print(customers_clean.isna().sum())


# ------------------------------------------------------------
# Products
# ------------------------------------------------------------

print("\nPRODUCTS")
print("Shape:", products_clean.shape)
print("Exact duplicates:", products_clean.duplicated().sum())
print("\nMissing values:")
print(products_clean.isna().sum())


# ------------------------------------------------------------
# Orders
# ------------------------------------------------------------

print("\nORDERS")
print("Shape:", orders_clean.shape)
print("Exact duplicates:", orders_clean.duplicated().sum())
print("\nMissing values:")
print(orders_clean.isna().sum())


# ------------------------------------------------------------
# Order Items
# ------------------------------------------------------------

print("\nORDER ITEMS")
print("Shape:", order_items_clean.shape)
print("Exact duplicates:", order_items_clean.duplicated().sum())
print("\nMissing values:")
print(order_items_clean.isna().sum())


# ------------------------------------------------------------
# Support Tickets
# ------------------------------------------------------------

print("\nSUPPORT TICKETS")
print(
    "Shape:",
    support_tickets_clean.shape
)

print(
    "Exact duplicates:",
    support_tickets_clean.duplicated().sum()
)

print("\nMissing values:")
print(
    support_tickets_clean.isna().sum()
)




# ============================================================
# 10. SAVE CLEANED DATASETS
# ============================================================

customers_clean.to_csv(
    os.path.join(
        cleaned_path,
        "customers_cleaned.csv"
    ),
    index=False
)

products_clean.to_csv(
    os.path.join(
        cleaned_path,
        "products_cleaned.csv"
    ),
    index=False
)

orders_clean.to_csv(
    os.path.join(
        cleaned_path,
        "orders_cleaned.csv"
    ),
    index=False
)

order_items_clean.to_csv(
    os.path.join(
        cleaned_path,
        "order_items_cleaned.csv"
    ),
    index=False
)

support_tickets_clean.to_csv(
    os.path.join(
        cleaned_path,
        "support_tickets_cleaned.csv"
    ),
    index=False
)


# ============================================================
# 11. CONFIRM FILES
# ============================================================

print("\n" + "=" * 70)
print("PHASE 6 COMPLETED — ALL FIVE DATASETS CLEANED")
print("=" * 70)

print("\nCleaned datasets saved to:")
print(cleaned_path)

print("\nFiles created:")

for file in sorted(os.listdir(cleaned_path)):
    print("✓", file)
```
